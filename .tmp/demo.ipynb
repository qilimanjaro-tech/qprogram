```python
from qprogram_qblox import QProgram
from qprogram import save, load, dumps
```

```python
qp = QProgram()
wait_duration = qp.variable("wait_duration", label="Wait Duration", units="ns")
with qp.for_loop(wait_duration, 0, 1_000, 10):
    qp.qblox.wait_trigger("drive", 1000)
    qp.wait("drive", wait_duration)
    qp.qblox.set_markers("abc", "123")
```

```python
save(qp, "./qblox.qp")
```

```python
qp2 = load("./qblox.qp")
```
